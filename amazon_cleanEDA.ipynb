```python
# STEP-2 DATA QUALITY REPORT

import pandas as pd
import numpy as np


#  Initialize a list to store report rows
report_rows = []

#  Identify numeric and boolean-like columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
bool_cols = ['is_prime_member', 'is_prime_eligible', 'is_festival_sale']  # example boolean columns
date_cols = ['order_date']  # columns expected to be dates

#  Generate report for each column
for col in df.columns:
    dtype = df[col].dtype
    non_null = df[col].notnull().sum()
    null_count = df[col].isnull().sum()
    null_percent = round((null_count / len(df)) * 100, 2)
    unique_vals = df[col].nunique()
    
    # Top 5 most frequent values for categorical columns
    if dtype == 'object' or dtype.name == 'category':
        top_values = df[col].value_counts().head(5).to_dict()
    else:
        top_values = None
    
    # Statistics for numeric columns
    if col in numeric_cols:
        mean = round(df[col].mean(), 2)
        min_val = df[col].min()
        max_val = df[col].max()
        std = round(df[col].std(), 2)
        # Outlier detection using IQR
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        outliers_low = df[df[col] < Q1 - 1.5 * IQR].shape[0]
        outliers_high = df[df[col] > Q3 + 1.5 * IQR].shape[0]
        total_outliers = outliers_low + outliers_high
    else:
        mean = min_val = max_val = std = None
        total_outliers = None
    
    # Invalid format detection for dates
    if col in date_cols:
        invalid_dates = df[pd.to_datetime(df[col], errors='coerce').isna()].shape[0]
    else:
        invalid_dates = None
    
    # Invalid entries for boolean-like columns
    if col in bool_cols:
        valid_bool = ['yes','no','true','false','1','0','y','n', np.nan, 0, 1]
        invalid_bool = df[~df[col].astype(str).str.lower().isin([str(v) for v in valid_bool])].shape[0]
    else:
        invalid_bool = None
    
    # Count duplicates for this column combination (example for transaction_id)
    if col == 'transaction_id':
        duplicates = df[col].duplicated().sum()
    else:
        duplicates = None
    
    # Add row to list
    report_rows.append({
        'Column': col,
        'DataType': dtype,
        'NonNullCount': non_null,
        'NullCount': null_count,
        'NullPercent': null_percent,
        'UniqueValues': unique_vals,
        'TopValues': top_values,
        'Mean': mean,
        'Min': min_val,
        'Max': max_val,
        'Std': std,
        'Outliers': total_outliers,
        'InvalidDates': invalid_dates,
        'InvalidBool': invalid_bool,
        'Duplicates': duplicates
    })

#  Convert list to DataFrame
report = pd.DataFrame(report_rows)

#  Save the report as CSV
report.to_csv("data_quality_report_N.csv", index=False)
print("Professional Data Quality Report generated and saved as 'data_quality_report_N.csv'")

#  Display first 20 rows (optional)
report.head(20)
```

```python
# FIXING THE DIFFERENT DATE FORMATS IN THE DATA

from dateutil.parser import parse

def safe_parse_date(x):
    """
    Safely parses dates from multiple inconsistent formats.
    Returns pandas.NaT for invalid or unrecognized dates.
    """
    try:
        return parse(x, dayfirst=True, fuzzy=True)
    except:
        return pd.NaT

# Clean separators first
df['order_date'] = (
    df['order_date']
    .astype(str)
    .str.strip()
    .str.replace(r'[-.]', '/', regex=True)
)

# Apply safe parsing (highly reliable)
df['order_date'] = df['order_date'].apply(safe_parse_date)

# Final standardized format
df['order_date'] = df['order_date'].dt.strftime('%Y-%m-%d')
```

```python
#HANDLING DUPLICATE TRANSACTIONS
import pandas as pd

def handle_transaction_duplicates(df, dup_keys=None, bulk_qty_threshold=5):
    """
    Handle duplicate transactions in an e-commerce dataset.
    
    Rules:
    - Duplicates are identified by dup_keys (customer_id, product_id, order_date, final_amount_inr)
    - Bulk orders (quantity > bulk_qty_threshold) are kept even if duplicates exist
    - Small duplicates (quantity <= bulk_qty_threshold) are likely data errors → keep one only
    - Logs removed duplicates for auditing
    
    Parameters:
    - df: pandas DataFrame
    - dup_keys: list of columns to identify duplicates
    - bulk_qty_threshold: int, min quantity to consider a bulk order
    
    Returns:
    - df_cleaned: cleaned DataFrame
    - removed_duplicates_log: DataFrame of removed duplicates
    """
    
    if dup_keys is None:
        dup_keys = ["customer_id", "product_id", "order_date", "final_amount_inr"]
    
    df = df.copy()
    
    # Identify potential duplicates
    potential_duplicates = df[df.duplicated(subset=dup_keys, keep=False)].copy()
    potential_duplicates["duplicate_flag"] = True
    
    # Flag bulk orders vs likely data errors
    potential_duplicates["bulk_order_flag"] = potential_duplicates["quantity"] > bulk_qty_threshold
    potential_duplicates["data_error_flag"] = potential_duplicates["quantity"] <= bulk_qty_threshold
    
    # Separate bulk orders (keep all)
    bulk_orders = potential_duplicates[potential_duplicates["bulk_order_flag"] == True]
    
    # Separate data errors (keep first only)
    data_errors = potential_duplicates[potential_duplicates["data_error_flag"] == True]
    data_errors_cleaned = data_errors.drop_duplicates(subset=dup_keys, keep="first")
    
    # Log removed duplicates (all but first of data errors)
    removed_duplicates_log = data_errors.drop_duplicates(subset=dup_keys, keep="first", inplace=False)
    
    # Remaining original dataset (excluding all potential duplicates)
    df_remaining = df[~df.index.isin(potential_duplicates.index)]
    
    # Combine bulk orders + cleaned data errors + remaining dataset
    df_cleaned = pd.concat([df_remaining, bulk_orders, data_errors_cleaned], ignore_index=True)
    
    # Optional: reset index
    df_cleaned.reset_index(drop=True, inplace=True)
    
    return df_cleaned, removed_duplicates_log
```

```python
import pandas as pd

def handle_transaction_duplicates_with_timestamp(df, dup_keys=None, bulk_qty_threshold=5, timestamp_col="order_timestamp"):
    """
    Handle duplicate transactions in an e-commerce dataset with timestamp validation.

    Rules:
    - Duplicates identified by dup_keys (customer_id, product_id, order_date, final_amount_inr)
    - Bulk orders (quantity > bulk_qty_threshold) are kept even if duplicates exist
    - Small duplicates (quantity <= bulk_qty_threshold) with same timestamp → likely data errors, keep one only
    - Transactions with same customer/product/date/amount but different timestamps → keep all
    - Logs removed duplicates for auditing

    Parameters:
    - df: pandas DataFrame
    - dup_keys: list of columns to identify duplicates
    - bulk_qty_threshold: int, min quantity to consider a bulk order
    - timestamp_col: column name for order timestamp

    Returns:
    - df_cleaned: cleaned DataFrame
    - removed_duplicates_log: DataFrame of removed duplicates
    """
    
    if dup_keys is None:
        dup_keys = ["customer_id", "product_id", "order_date", "final_amount_inr"]
    
    df = df.copy()
    
    # Ensure timestamp is datetime
    df[timestamp_col] = pd.to_datetime(df[timestamp_col], errors='coerce')
    
    # Identify potential duplicates by key columns
    potential_duplicates = df[df.duplicated(subset=dup_keys, keep=False)].copy()
    potential_duplicates["duplicate_flag"] = True
    
    # Flag bulk orders vs small duplicates
    potential_duplicates["bulk_order_flag"] = potential_duplicates["quantity"] > bulk_qty_threshold
    potential_duplicates["small_duplicate_flag"] = potential_duplicates["quantity"] <= bulk_qty_threshold
    
    # Bulk orders → keep all
    bulk_orders = potential_duplicates[potential_duplicates["bulk_order_flag"] == True]
    
    # Small duplicates → check timestamp
    small_duplicates = potential_duplicates[potential_duplicates["small_duplicate_flag"] == True]
    
    # Remove exact timestamp duplicates (keep first)
    if timestamp_col in small_duplicates.columns:
        data_errors_cleaned = small_duplicates.drop_duplicates(subset=dup_keys + [timestamp_col], keep="first")
        removed_duplicates_log = small_duplicates.drop_duplicates(subset=dup_keys + [timestamp_col], keep="first", inplace=False)
    else:
        # fallback if timestamp not present
        data_errors_cleaned = small_duplicates.drop_duplicates(subset=dup_keys, keep="first")
        removed_duplicates_log = small_duplicates.drop_duplicates(subset=dup_keys, keep="first", inplace=False)
    
    # Remaining original dataset (excluding all potential duplicates)
    df_remaining = df[~df.index.isin(potential_duplicates.index)]
    
    # Combine all parts
    df_cleaned = pd.concat([df_remaining, bulk_orders, data_errors_cleaned], ignore_index=True)
    df_cleaned.reset_index(drop=True, inplace=True)
    
    return df_cleaned, removed_duplicates_log
```

```python
import pandas as pd

def clean_orders_with_bulk_and_transaction_duplicates(df, bulk_qty_threshold=5):
    """
    Handle both duplicate transactions and bulk orders in e-commerce dataset.

    Rules:
    1. Remove duplicate transaction_ids (data errors)
    2. Identify potential duplicates by customer_id, product_id, order_date, final_amount_inr
       - Bulk orders (quantity > bulk_qty_threshold) → keep all
       - Small duplicates (quantity <= bulk_qty_threshold) → keep first only
    3. Logs all removed duplicates for auditing

    Parameters:
    - df: pandas DataFrame
    - bulk_qty_threshold: int, minimum quantity to consider a bulk order

    Returns:
    - df_cleaned: cleaned DataFrame
    - removed_transaction_duplicates: DataFrame of removed transaction_id duplicates
    - removed_small_duplicates: DataFrame of removed small duplicate orders
    """
    
    df = df.copy()
    
    # Remove duplicate transaction_ids
    duplicates_txn = df[df.duplicated(subset=["transaction_id"], keep=False)]
    removed_transaction_duplicates = duplicates_txn[duplicates_txn.duplicated(subset=["transaction_id"], keep="first")].copy()
    df = df.drop_duplicates(subset=["transaction_id"], keep="first").reset_index(drop=True)
    
    # Handle bulk orders vs small duplicates
    dup_keys = ["customer_id", "product_id", "order_date", "final_amount_inr"]
    potential_duplicates = df[df.duplicated(subset=dup_keys, keep=False)].copy()
    potential_duplicates["bulk_order_flag"] = potential_duplicates["quantity"] > bulk_qty_threshold
    potential_duplicates["small_duplicate_flag"] = potential_duplicates["quantity"] <= bulk_qty_threshold
    
    # Bulk orders → keep all
    bulk_orders = potential_duplicates[potential_duplicates["bulk_order_flag"] == True]
    
    # Small duplicates → keep first only
    small_duplicates = potential_duplicates[potential_duplicates["small_duplicate_flag"] == True]
    small_duplicates_cleaned = small_duplicates.drop_duplicates(subset=dup_keys, keep="first")
    
    # Log removed small duplicates
    removed_small_duplicates = small_duplicates.drop_duplicates(subset=dup_keys, keep="first", inplace=False)
    
    # Remaining dataset (excluding all potential duplicates)
    df_remaining = df[~df.index.isin(potential_duplicates.index)]
    
    # Combine everything
    df_cleaned = pd.concat([df_remaining, bulk_orders, small_duplicates_cleaned], ignore_index=True)
    df_cleaned.reset_index(drop=True, inplace=True)
    
    return df_cleaned, removed_transaction_duplicates, removed_small_duplicates
```

```python
#-----------------------------------
#EDA 3- CUSTOMER SEGMENTATION - RFM
#-----------------------------------
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. Calculate Recency, Frequency, Monetary for each customer

# Convert order_date to datetime 
df_clean['order_date'] = pd.to_datetime(df_clean['order_date'])

# Define analysis date as one day after the max order date 
snapshot_date = df_clean['order_date'].max() + pd.Timedelta(days=1)

# Aggregate metrics
rfm = df_clean.groupby('customer_id').agg({
    'order_date': lambda x: (snapshot_date - x.max()).days,  # Recency
    'transaction_id': 'nunique',                             # Frequency
    'final_amount_inr_cleaned': 'sum'                        # Monetary
}).reset_index()

rfm.rename(columns={
    'order_date': 'Recency',
    'transaction_id': 'Frequency',
    'final_amount_inr_cleaned': 'Monetary'
}, inplace=True)

# 2. Create RFM quartiles for scoring (1-4 scale, 4 is best)
# For Recency, lower days means better, so invert score

rfm['R_Score'] = pd.qcut(rfm['Recency'], 4, labels=[4,3,2,1]).astype(int)
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 4, labels=[1,2,3,4]).astype(int)
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 4, labels=[1,2,3,4]).astype(int)

# Combine RFM score
rfm['RFM_Score'] = rfm['R_Score'].map(str) + rfm['F_Score'].map(str) + rfm['M_Score'].map(str)

# 3. Define simple segments based on RFM score

def segment_customer(row):
    if row['R_Score'] >= 3 and row['F_Score'] >= 3 and row['M_Score'] >= 3:
        return 'Champions'
    elif row['R_Score'] >= 3 and row['F_Score'] >= 2:
        return 'Loyal Customers'
    elif row['R_Score'] >= 3 and row['F_Score'] < 2:
        return 'Potential Loyalist'
    elif row['R_Score'] == 4 and row['F_Score'] <= 2:
        return 'Recent Customers'
    elif row['R_Score'] <= 2 and row['F_Score'] >= 3:
        return 'At Risk'
    else:
        return 'Others'

rfm['Segment'] = rfm.apply(segment_customer, axis=1)

# 4. Visualize with scatter plots

plt.figure(figsize=(12, 6))

# Scatter plot Frequency vs Monetary colored by segment
sns.scatterplot(data=rfm, x='Frequency', y='Monetary', hue='Segment', palette='Set2', alpha=0.7)
plt.title('Customer Segmentation: Frequency vs Monetary Value')
plt.xlabel('Frequency (Number of Transactions)')
plt.ylabel('Monetary Value (Total Spend INR)')
plt.legend(loc='upper right')
plt.show()

plt.figure(figsize=(12, 6))

# Scatter plot Recency vs Monetary colored by segment
sns.scatterplot(data=rfm, x='Recency', y='Monetary', hue='Segment', palette='Set2', alpha=0.7)
plt.title('Customer Segmentation: Recency vs Monetary Value')
plt.xlabel('Recency (Days Since Last Purchase)')
plt.ylabel('Monetary Value (Total Spend INR)')
plt.legend(loc='upper right')
plt.show()

# Optional: Summary counts per segment
print(rfm['Segment'].value_counts())
```

```python
#---------------------------------
#EDA 14- CLV - COHORT ANALYSIS
#-----------------------------------
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Convert order_date to datetime
df_clean['order_date'] = pd.to_datetime(df_clean['order_date'])

# Create Cohorts: based on the month of first purchase
df_clean['cohort_month'] = df_clean.groupby('customer_id')['order_date'].transform('min').dt.to_period('M')
df_clean['order_month'] = df_clean['order_date'].dt.to_period('M')

#  Calculate Cohort Index: months since first purchase
def cohort_index(df):
    return (df.order_month.dt.year - df.cohort_month.dt.year)*12 + (df.order_month.dt.month - df.cohort_month.dt.month) + 1

df_clean['cohort_index'] = cohort_index(df_clean)

# Retention Matrix
cohort_data = df_clean.groupby(['cohort_month','cohort_index'])['customer_id'].nunique().reset_index()
cohort_pivot = cohort_data.pivot(index='cohort_month', columns='cohort_index', values='customer_id')
cohort_size = cohort_pivot.iloc[:,0]
retention = cohort_pivot.divide(cohort_size, axis=0)

# Plot Retention Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(retention, annot=True, fmt=".0%", cmap="YlGnBu")
plt.title('Customer Retention Rate by Cohort')
plt.xlabel('Months since First Purchase')
plt.ylabel('Cohort Month')
plt.show()

# Customer Lifetime Value (CLV)
# Calculate monetary value per customer
clv_data = df_clean.groupby('customer_id').agg({
    'final_amount_inr_cleaned':'sum',
    'order_date':'count'
}).rename(columns={'final_amount_inr_cleaned':'monetary_value','order_date':'frequency'})

# Estimate average order value (AOV) and repeat purchase rate
clv_data['avg_order_value'] = clv_data['monetary_value'] / clv_data['frequency']

# Optional: assume simple retention rate model
# CLV = Avg Order Value * Frequency * Profit Margin (simplified)
profit_margin = 0.2
clv_data['CLV'] = clv_data['monetary_value'] * profit_margin

# CLV Distribution Plot
plt.figure(figsize=(10,6))
sns.histplot(clv_data['CLV'], bins=50, kde=True)
plt.title('Customer Lifetime Value (CLV) Distribution')
plt.xlabel('CLV (INR)')
plt.ylabel('Number of Customers')
plt.show()

# CLV by Acquisition Year
df_clean['acquisition_year'] = df_clean.groupby('customer_id')['order_date'].transform('min').dt.year
clv_by_year = clv_data.merge(df_clean[['customer_id','acquisition_year']].drop_duplicates(), on='customer_id')
clv_year_summary = clv_by_year.groupby('acquisition_year')['CLV'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.barplot(data=clv_year_summary, x='acquisition_year', y='CLV', palette="Blues_d")
plt.title('Average CLV by Acquisition Year')
plt.xlabel('Acquisition Year')
plt.ylabel('Average CLV (INR)')
plt.show()
```

```python
#------------------------------------
#EDA 17- CUSTOMER JOURNEY ANALYSIS
#-----------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go

# Ensure order_date is datetime
df_clean['order_date'] = pd.to_datetime(df_clean['order_date'])


#  Customer Stage Assignment

# First purchase date
df_clean['first_purchase_date'] = df_clean.groupby('customer_id')['order_date'].transform('min')

# Customer metrics
customer_journey = df_clean.groupby('customer_id').agg(
    first_purchase=('first_purchase_date','min'),
    last_purchase=('order_date','max'),
    frequency=('transaction_id','count'),
    total_spent=('final_amount_inr_cleaned','sum')
).reset_index()

# Define customer stage
def customer_stage(freq):
    if freq == 1:
        return 'New'
    elif freq <= 3:
        return 'Engaged'
    elif freq <= 10:
        return 'Loyal'
    else:
        return 'Champion'

customer_journey['stage'] = customer_journey['frequency'].apply(customer_stage)

# --------------------------
# Purchase Frequency Distribution
# --------------------------

plt.figure(figsize=(10,5))
sns.histplot(customer_journey['frequency'], bins=30, color='skyblue', log_scale=True)
plt.title("Customer Purchase Frequency Distribution")
plt.xlabel("Number of Orders")
plt.ylabel("Number of Customers (log scale)")
plt.show()

# --------------------------
# Category Transition Analysis
# --------------------------

# Sort and get first 3 purchases per customer
df_sorted = df_clean.sort_values(['customer_id','order_date'])
df_first3 = df_sorted.groupby('customer_id').head(3)

# Prepare transitions
df_first3['prev_category'] = df_first3.groupby('customer_id')['subcategory'].shift(0)
df_first3['next_category'] = df_first3.groupby('customer_id')['subcategory'].shift(-1)
transitions = df_first3.dropna(subset=['next_category'])
transition_matrix = pd.crosstab(transitions['prev_category'], transitions['next_category'], normalize='index')

# Sankey diagram setup
source_labels = transition_matrix.index.tolist()
target_labels = transition_matrix.columns.tolist()
all_labels = list(set(source_labels + target_labels))

source_indices = [all_labels.index(s) for s in transition_matrix.index.repeat(len(transition_matrix.columns))]
target_indices = [all_labels.index(t) for t in transition_matrix.columns.tolist()*len(transition_matrix.index)]
values = transition_matrix.values.flatten()

fig_sankey = go.Figure(data=[go.Sankey(
    node=dict(label=all_labels, pad=15, thickness=20),
    link=dict(source=source_indices, target=target_indices, value=values)
)])
fig_sankey.update_layout(title_text="Customer Category Transition Sankey Diagram", font_size=12)
fig_sankey.show()

# --------------------------
# Customer Evolution Timeline
# --------------------------

df_clean['year'] = df_clean['order_date'].dt.year
evolution = df_clean.groupby(['year','customer_id']).size().groupby(level=0).count().reset_index(name='active_customers')

plt.figure(figsize=(12,6))
sns.lineplot(data=evolution, x='year', y='active_customers', marker='o', color='orange')
plt.title("Customer Evolution: Active Customers per Year")
plt.ylabel("Number of Active Customers")
plt.xlabel("Year")
plt.grid(True)
plt.show()

# --------------------------
#Display Customer Stage Distribution
# --------------------------

plt.figure(figsize=(8,5))
sns.countplot(data=customer_journey, x='stage', order=['New','Engaged','Loyal','Champion'], palette='Set2')
plt.title("Customer Stage Distribution")
plt.xlabel("Stage")
plt.ylabel("Number of Customers")
plt.show()
```
